**Inline Question 2**

Although gradcheck is reliable softmax loss, it is possible that for SVM loss, once in a while, a dimension in the gradcheck will not match exactly. What could such a discrepancy be caused by? Is it a reason for concern? What is a simple example in one dimension where a svm loss gradient check could fail? How would change the margin affect of the frequency of this happening?

Note that SVM loss for a sample $(x_i, y_i)$ is defined as: $$L_i = \sum_{j\ne y_i}\max(0, s_j - s_{y_i} + \Delta)$$ where $j$ iterates over all classes except the correct class $y_i$ and $s_j$ denotes the classifier score for $j^{th}$ class. $\Delta$ is a scalar margin. For more information, refer to 'Multiclass Support Vector Machine loss' on [this](https://cs231n.github.io/linear-classify/) page.

*Hint: the SVM loss function is not strictly speaking differentiable.*


**Answer:** The hinge loss has a kink at zero margin violation. A central finite difference can cross that kink, while the analytic implementation chooses one branch (or a subgradient at the kink). For f(t)=max(0,t), the central difference at t=0 is 1/2, whereas an implementation may return 0. An isolated discrepancy near a kink need not indicate a bug; persistent discrepancies away from kinks do. Changing Delta moves the kink relative to the score differences, changing how often the finite-difference interval crosses it. There is no universal monotonic relationship between Delta and failure frequency; it depends on the score distribution and finite-difference step.

### Stochastic Gradient Descent

We now have vectorized and efficient expressions for the loss, the gradient and our gradient matches the numerical gradient. We are therefore ready to do SGD to minimize the loss. Your code for this part will be written inside `cs231n/classifiers/linear_classifier.py`.

```python
# Save the trained model for autograder.
softmax.save("softmax.npy")
```

```python
# Select hyperparameters using validation accuracy only.
learning_rates = [1e-7, 3e-7, 1e-6]
regularization_strengths = [1e3, 5e3, 1e4, 2.5e4]
results = {}
best_val = -1
best_softmax = None
for lr in learning_rates:
    for reg in regularization_strengths:
        np.random.seed(42)
        candidate = Softmax()
        candidate.train(X_train, y_train, learning_rate=lr, reg=reg,
                        num_iters=2000, batch_size=200)
        train_acc = np.mean(candidate.predict(X_train) == y_train)
        val_acc = np.mean(candidate.predict(X_val) == y_val)
        results[(lr, reg)] = (train_acc, val_acc)
        print('lr %e reg %e train accuracy: %f val accuracy: %f' %
              (lr, reg, train_acc, val_acc), flush=True)
        if val_acc > best_val:
            best_val = val_acc
            best_softmax = candidate
print('best validation accuracy achieved: %f' % best_val)
assert best_val > 0.36, 'Further validation-set tuning is needed.'
```

```python
# Visualize the cross-validation results
import math
import pdb

# pdb.set_trace()

x_scatter = [math.log10(x[0]) for x in results]
y_scatter = [math.log10(x[1]) for x in results]

# plot training accuracy
marker_size = 100
colors = [results[x][0] for x in results]
plt.subplot(2, 1, 1)
plt.tight_layout(pad=3)
plt.scatter(x_scatter, y_scatter, marker_size, c=colors, cmap=plt.cm.coolwarm)
plt.colorbar()
plt.xlabel('log learning rate')
plt.ylabel('log regularization strength')
plt.title('CIFAR-10 training accuracy')

# plot validation accuracy
colors = [results[x][1] for x in results] # default size of markers is 20
plt.subplot(2, 1, 2)
plt.scatter(x_scatter, y_scatter, marker_size, c=colors, cmap=plt.cm.coolwarm)
plt.colorbar()
plt.xlabel('log learning rate')
plt.ylabel('log regularization strength')
plt.title('CIFAR-10 validation accuracy')
plt.show()
```

```python
# Evaluate the best softmax on test set
y_test_pred = best_softmax.predict(X_test)
test_accuracy = np.mean(y_test == y_test_pred)
print('Softmax classifier on raw pixels final test set accuracy: %f' % test_accuracy)
```

**Inline Question 4** - *True or False*

Suppose the overall training loss is defined as the sum of the per-datapoint loss over all training examples. It is possible to add a new datapoint to a training set that would change the softmax loss, but leave the SVM loss unchanged.

**Answer:** True, for fixed weights and the stated sum of per-example losses. Add a point whose correct-class score exceeds every incorrect-class score by at least the SVM margin. Its hinge loss is zero, so the total SVM loss is unchanged. For finite scores, softmax assigns the correct class probability strictly below one, so this point contributes positive cross-entropy and increases the total softmax loss. The regularization term is unchanged at fixed weights.
